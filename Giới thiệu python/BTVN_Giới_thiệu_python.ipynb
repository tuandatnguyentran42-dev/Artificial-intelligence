```python
# BTVN Bài 1
def tron_mang(a, b):
    c = []
    min_len = min(len(a), len(b))

    for i in range(min_len):
        c.append(min(a[i], b[i]))

    if len(a) > len(b):
        c.extend(a[min_len:])
    else:
        c.extend(b[min_len:])

    return c


a = list(map(int, input("Nhap mang a: ").split()))
b = list(map(int, input("Nhap mang b: ").split()))

c = tron_mang(a, b)

print("Mang ket qua:", c)
```

```python
# BTVN Bài 2
def nhap_ma_tran(m, n):
    a = []
    for i in range(m):
        hang = list(map(int, input().split()))
        a.append(hang)
    return a


def xuat_ma_tran(a):
    for hang in a:
        for x in hang:
            print(f"{x:4}", end="")
        print()


# a
def tong_tam_giac_tren_cheo_phu(a, n):
    tong = 0
    for i in range(n):
        for j in range(n):
            if i + j <= n - 1:
                tong += a[i][j]
    return tong


# b
def doi_am_thanh_duong(a):
    for i in range(len(a)):
        for j in range(len(a[i])):
            if a[i][j] < 0:
                a[i][j] = abs(a[i][j])


# c
def thay_so_chan(a, x):
    for i in range(len(a)):
        for j in range(len(a[i])):
            if a[i][j] % 2 == 0:
                a[i][j] = x


# d
def kiem_tra_toan_chan(a):
    for hang in a:
        for x in hang:
            if x % 2 != 0:
                return False
    return True


# e
def kiem_tra_doi_xung(a, n):
    for i in range(n):
        for j in range(n):
            if a[i][j] != a[j][i]:
                return False
    return True


# f
def cheo_chinh_tang(a, n):
    for i in range(n - 1):
        if a[i][i] >= a[i + 1][i + 1]:
            return False
    return True


# g
def xuat_tam_giac_duoi_cheo_phu(a, n):
    for i in range(n):
        for j in range(n):
            if i + j >= n - 1:
                print(f"{a[i][j]:4}", end="")
        print()


# h
def cheo_phu_giam(a, n):
    for i in range(n - 1):
        if a[i][n - 1 - i] <= a[i + 1][n - 2 - i]:
            return False
    return True


# ========== Main ==========
m = int(input("Nhập số dòng m: "))
n = int(input("Nhập số cột n: "))

print("Nhập ma trận:")
a = nhap_ma_tran(m, n)

print("\n=== Ma trận ban đầu ===")
xuat_ma_tran(a)

print("\n========== Kết quả ==========")

if m == n:
    print("\nTổng tam giác trên chéo phụ:", tong_tam_giac_tren_cheo_phu(a, n))

doi_am_thanh_duong(a)
print("\n=== Sau khi đổi số âm ===")
xuat_ma_tran(a)

x = int(input("\nNhập x: "))
thay_so_chan(a, x)
print("\n=== Sau khi thay số chẵn ===")
xuat_ma_tran(a)

print("\nToàn chẵn:", kiem_tra_toan_chan(a))

if m == n:
    print("\nĐối xứng:", kiem_tra_doi_xung(a, n))
    print("\nChéo chính tăng:", cheo_chinh_tang(a, n))

    print("\n=== Tam giác dưới chéo phụ ===")
    xuat_tam_giac_duoi_cheo_phu(a, n)

    print("\nChéo phụ giảm:", cheo_phu_giam(a, n))
```
