```python
# Bài 13
def tron_mang(a, b):
    c = []
    min_len = min(len(a), len(b))

    for i in range(min_len):
        c.append(a[i] + b[i])

    if len(a) > len(b):
        c.extend(a[min_len:])
    else:
        c.extend(b[min_len:])

    return c


a = list(map(int, input("Nhap mang a: ").split()))
b = list(map(int, input("Nhap mang b: ").split()))

c = tron_mang(a, b)

print("Mang ket qua:", c)
```

```python
# Bài 14
# a.
import random

def tao_mang():
    m = int(input("Nhập số dòng m: "))
    n = int(input("Nhập số cột n: "))

    a = []
    for i in range(m):
        row = []
        for j in range(n):
            row.append(random.randint(1, 100))
        a.append(row)

    return a

def xuat_mang(a):
    for row in a:
        print(row)

# b.
def xuat_dong_k(a, k):
    if 0 <= k < len(a):
        return a[k]
    return "Dòng không hợp lệ"

# c.
def xuat_cot_k(a, k):
    if 0 <= k < len(a[0]):
        return [a[i][k] for i in range(len(a))]
    return "Cột không hợp lệ"

# d.
def dong_tong_lon_nhat_45(a):
    max_sum = -1
    index = -1
    for i in range(len(a)):
        s = sum(a[i])
        if s >= 45 and s > max_sum:
            max_sum = s
            index = i
    return index, max_sum

# e.
def cot_tich_nho_nhat(a):
    n = len(a[0])
    min_tich = None
    index = -1
    for j in range(n):
        tich = 1
        for i in range(len(a)):
            tich *= a[i][j]
        if min_tich is None or tich < min_tich:
            min_tich = tich
            index = j
    return index, min_tich

# f.
def dong_chan_cot_le(a):
    kq = []
    for i in range(len(a)):
        if i % 2 == 0:
            for j in range(len(a[0])):
                if j % 2 == 1:
                    kq.append(a[i][j])
    return kq

# g. TBC phần tử chẵn thuộc dòng lẻ
def tbc_chan_dong_le(a):
    tong = 0
    dem = 0
    for i in range(len(a)):
        if i % 2 == 1:
            for j in range(len(a[0])):
                if a[i][j] % 2 == 0:
                    tong += a[i][j]
                    dem += 1
    return tong / dem if dem != 0 else 0

# h.
def tbc_bien(a):
    m = len(a)
    n = len(a[0])
    tong = 0
    dem = 0

    for i in range(m):
        for j in range(n):
            if i == 0 or i == m-1 or j == 0 or j == n-1:
                tong += a[i][j]
                dem += 1
    return tong / dem

# i.
def tbc_tich_khong_bien(a):
    m = len(a)
    n = len(a[0])
    tich = 1
    dem = 0

    for i in range(1, m-1):
        for j in range(1, n-1):
            tich *= a[i][j]
            dem += 1

    return tich / dem if dem != 0 else 0

# ========== Main ==========

a = tao_mang()

print("\nMa trận:")
xuat_mang(a)

k = int(input("\nNhập k: "))
print("Dòng k:", xuat_dong_k(a, k))
print("Cột k:", xuat_cot_k(a, k))

print("\nDòng có tổng >=45 lớn nhất:", dong_tong_lon_nhat_45(a))
print("Cột có tích nhỏ nhất:", cot_tich_nho_nhat(a))

print("\nPhần tử dòng chẵn, cột lẻ:", dong_chan_cot_le(a))
print("Trung bình cộng chẵn dòng lẻ:", tbc_chan_dong_le(a))

print("Trung bình cộng phần tử biên:", tbc_bien(a))
print("Trung bình tích không biên:", tbc_tich_khong_bien(a))
```

```python
# Bài 15
# a.
class SinhVien:
    def __init__(self, ma, ten, nam_sinh, diem_tb):
        self.ma = ma
        self.ten = ten
        self.nam_sinh = nam_sinh
        self.diem_tb = diem_tb

    def __str__(self):
        return f"{self.ma} | {self.ten} | {self.nam_sinh} | {self.diem_tb:.1f}"

# b.
def nhap_danh_sach():
    ds = []
    n = int(input("Nhập số sinh viên: "))
    for i in range(n):
        print(f"\n========== Sinh viên {i+1} ==========")
        ma = input("Mã SV (10 ký tự): ")
        ten = input("Tên (<=20 ký tự): ")
        nam_sinh = int(input("Năm sinh: "))
        diem_tb = float(input("Điểm trung bình: "))
        ds.append(SinhVien(ma, ten, nam_sinh, diem_tb))
        print("=================================")
    return ds

# c.
def dem_len_lop(ds):
    return sum(1 for sv in ds if sv.diem_tb >= 5)

# d.
def sv_20_tuoi(ds):
    from datetime import datetime
    nam_hien_tai = datetime.now().year

    print("\nSinh viên đủ 20 tuổi:")
    for sv in ds:
        if nam_hien_tai - sv.nam_sinh >= 20:
            print(f"{sv.ma} | {sv.ten} | {nam_hien_tai - sv.nam_sinh} tuổi")

# e.
def dem_sv_dh(ds):
    return sum(1 for sv in ds if sv.ma[1:3] == "DH")

# f.
def dem_ten_lan(ds):
    return sum(1 for sv in ds if sv.ten.split()[-1] == "Lan")

# h.
def dem_ho_phan(ds):
    return sum(1 for sv in ds if sv.ten.split()[0] == "Phan")


# ========== Main ==========
ds = nhap_danh_sach()

print("\nDanh sách sinh viên:")
for sv in ds:
    print(sv)

print("\nSố sinh viên đủ điều kiện lên lớp:", dem_len_lop(ds))

sv_20_tuoi(ds)

print("\nSố sinh viên hệ DH:", dem_sv_dh(ds))
print("\nSố sinh viên tên Lan:", dem_ten_lan(ds))
print("\nSố sinh viên họ Phan:", dem_ho_phan(ds))
```
