```python
import numpy as np
import matplotlib.pyplot as plt
import math

print("--- BAI 1: TRUC QUAN HOA ---")

def f(x):
    return x**3 - np.log(x) - 3

x_vals = np.linspace(0.1, 3, 100)
y_vals = f(x_vals)

plt.figure(figsize=(8, 5))
plt.plot(x_vals, y_vals, label='f(x) = x^3 - ln(x) - 3', color='blue')
plt.axhline(0, color='red', linestyle='--', label='y=0')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.legend()
plt.show()

print("Dua vao do thi, khoang phan ly nghiem duy nhat (do dai 1) la:")
print("Khoang: [1, 2]")
print("\n")
```

```python

# BAI 2
print("--- BAI 2: PHUONG PHAP CHIA DOI ---")

def bisection_method(f, a, b, tol=1e-5):
    if f(a) * f(b) >= 0:
        print("Loi: f(a) va f(b) phai trai dau!")
        return None, 0

    iterations = 0
    c = a

    while (b - a) / 2.0 > tol:
        c = (a + b) / 2.0
        if f(c) == 0:
            break
        elif f(a) * f(c) < 0:
            b = c
        else:
            a = c
        iterations += 1

    return c, iterations


def f(x):
    import numpy as np
    return x**3 - np.log(x) - 3


root_bisect, iter_bisect = bisection_method(f, 1, 2, tol=1e-5)

if root_bisect is not None:
    print(f"Nghiem duong duy nhat (chia doi): x = {root_bisect:.6f}")
    print(f"So lan lap: {iter_bisect}\n")
else:
    print("Khong tim thay nghiem trong khoang da cho hoac co loi xay ra.")
```

```python

import numpy as np

# ma tran he so
A = np.array([
    [12, 1, -2],
    [3, 14, -1],
    [-2, 2, 15]
], dtype=float)

# vector tu do
b = np.array([7.5, 8.7, 9.2], dtype=float)

# sai so
eps = 1e-4

#Jacobi
def jacobi(A, b, eps):
    n = len(b)
    x = np.zeros(n)
    x_new = np.zeros(n)

    while True:
        for i in range(n):
            s = 0
            for j in range(n):
                if j != i:
                    s += A[i][j] * x[j]
            x_new[i] = (b[i] - s) / A[i][i]

        # tinh sai so vo cung
        err = np.max(np.abs(x_new - x))
        if err < eps:
            break

        x = x_new.copy()

    return x_new

#Gauss-Seidel
def gauss_seidel(A, b, eps):
    n = len(b)
    x = np.zeros(n)

    while True:
        x_old = x.copy()

        for i in range(n):
            s1 = sum(A[i][j] * x[j] for j in range(i))
            s2 = sum(A[i][j] * x_old[j] for j in range(i+1, n))
            x[i] = (b[i] - s1 - s2) / A[i][i]

        # sai so
        err = np.max(np.abs(x - x_old))
        if err < eps:
            break

    return x

# ================== Chay ==================
x_jacobi = jacobi(A, b, eps)
x_gs = gauss_seidel(A, b, eps)

print("Nghiem Jacobi:", x_jacobi)
print("Nghiem Gauss-Seidel:", x_gs)
```
