```python
import random

teacher = {
    "ThayAn":"Toan",
    "CoBinh":"Toan",
    "ThayCuong":"Tin",
    "CoDung":"Tin"
}

classes = ["10A1","10A2","11A1"]

days = ["Thu2","Thu3","Thu4","Thu5","Thu6"]

periods = [1,2,3]

subjectNeed = {
    "Toan":4,
    "Tin":2
}

lesson = []

for c in classes:

    for i in range(subjectNeed["Toan"]):
        lesson.append([c,"Toan"])

    for i in range(subjectNeed["Tin"]):
        lesson.append([c,"Tin"])

populationSize = 20

generation = 100

def createIndividual():

    individual = []

    for l in lesson:

        c = l[0]
        s = l[1]

        possibleTeacher = []

        for t in teacher:
            if teacher[t] == s:
                possibleTeacher.append(t)

        gv = random.choice(possibleTeacher)

        d = random.choice(days)
        p = random.choice(periods)

        individual.append([c,s,gv,d,p])

    return individual

def fitness(individual):

    penalty = 0

    teacherCheck = []
    classCheck = []

    for x in individual:

        c = x[0]
        gv = x[2]
        d = x[3]
        p = x[4]

        if [gv,d,p] in teacherCheck:
            penalty += 10
        else:
            teacherCheck.append([gv,d,p])

        if [c,d,p] in classCheck:
            penalty += 10
        else:
            classCheck.append([c,d,p])

    return 1/(1+penalty)

def crossover(p1,p2):

    point = random.randint(1,len(lesson)-1)

    child1 = p1[:point] + p2[point:]
    child2 = p2[:point] + p1[point:]

    return child1,child2

def mutation(individual):

    for i in range(len(individual)):

        if random.random() < 0.1:

            c = individual[i][0]
            s = individual[i][1]

            possibleTeacher = []

            for t in teacher:
                if teacher[t] == s:
                    possibleTeacher.append(t)

            gv = random.choice(possibleTeacher)

            d = random.choice(days)
            p = random.choice(periods)

            individual[i] = [c,s,gv,d,p]

    return individual

population = []

for i in range(populationSize):
    population.append(createIndividual())

print("\n=========== QUA TRINH TIEN HOA ===========\n")

print(
    "{:<10} {:<10}".format(
        "THE HE",
        "FITNESS"
    )
)

print("-"*22)

for gen in range(generation):

    population = sorted(
        population,
        key=lambda x: fitness(x),
        reverse=True
    )

    newPopulation = population[:10]

    while len(newPopulation) < populationSize:

        p1 = random.choice(population[:10])
        p2 = random.choice(population[:10])

        c1,c2 = crossover(p1,p2)

        c1 = mutation(c1)
        c2 = mutation(c2)

        newPopulation.append(c1)

        if len(newPopulation) < populationSize:
            newPopulation.append(c2)

    population = newPopulation

    if (gen+1)%10 == 0:

        print(
            "{:<10} {:<10}".format(
                gen+1,
                round(fitness(population[0]),2)
            )
        )

best = population[0]

print("\n================ THOI KHOA BIEU TOT NHAT ================\n")

print(
    "{:<8} {:<8} {:<12} {:<8} {:<5}".format(
        "LOP",
        "MON",
        "GIAO VIEN",
        "NGAY",
        "TIET"
    )
)

print("-"*45)

for x in best:

    print(
        "{:<8} {:<8} {:<12} {:<8} {:<5}".format(
            x[0],
            x[1],
            x[2],
            x[3],
            x[4]
        )
    )
```
